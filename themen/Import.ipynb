Es gibt verschiedene Möglichkeiten, Module in Python zu importieren.

Am Beispiel von https://docs.python.org/3/library/math.html

## Import aller Funktionen (nicht empfohlen)

```
from math import *

print("Fakultät von 5:", factorial(5))
```

## Selektiver Import

```
from math import sin, radians  
  
winkel = 45  
bogenmass = radians(winkel)  
  
print("sin({}°) == sin({})".format(winkel, bogenmass))
```

## Import ohne Alias

```
import math

print("π =", math.pi)
print("e =", math.e)
```

## Import mit Alias (empfohlen)

```
import math as m

print("Wurzel von 16:", m.sqrt(16))
```
