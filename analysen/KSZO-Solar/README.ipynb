# Rohdaten einlesen

Wir benötigen die Bibliothek `pandas`.

```python
import pandas as pd
```

Somit können wir die Rohdaten mit dem Befehl `read_csv` bequem einlesen.

## csv-Datei

Wir entpacken das komprimierte zip-Archiv `KSZO-solar-data.csv.zip` zuerst

```python
%%bash
unzip -o KSZO-solar-data.csv.zip
```

und können anschliessend den Inhalt der Datei `KSZO-solar-data.csv` untersuchen. Dazu öffnen wir die Datei per Doppelklick. Achte auf den `Dateinamen`.

Wir lesen die entpackte Datei `KSZO-solar-data.csv` in einen DataFrame mit dem Namen `df1` ein:

```python
df1 = pd.read_csv('KSZO-solar-data.csv', sep=",")
```

und schauen uns den Inhalt von `df1` an

```python
df1
```

## csv-Datei-zip-Archiv

Wenn sehr grosse Datenmengen untersucht werden sollen, bietet es sich an, das .zip-Archiv direkt einzulesen.

```python
df2 = pd.read_csv('KSZO-solar-data.csv.zip', sep=",")
```

```python
df2
```

Sind die beiden DataFrames wirklich identisch?

```python
assert df1.equals(df2)
```
