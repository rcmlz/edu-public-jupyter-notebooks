### Hilfsfunktionen

```python
def show_models_from_(output):
    """Zeigt die Modelle in einer ollama-python-client-Ausgabe an"""
    for entry in output['models']:
        print(entry["model"])
```

Für die einfachere Handhabung nutzen wir die zwei Funktionen `ask()` und `chat()`. Unser Code im Notebook ist dann übersichtlicher.

Wichtig: Die globalen Variablen `model_name` und `messages` müssen vor der Verwendung definiert werden. Dazu am besten alle Code-Zellen in diesem Abschnitt einmalig nacheinander ausführen. 

```python
from IPython.display import display, Markdown

def ask(prompt):
    """Einfacher Chat-Bot mit global spezifiziertem Client, Modell - ohne Historie"""
    global client
    global model_name
    response = client.generate(model=model_name, prompt=prompt)
    display(Markdown(response.response))

def chat(prompt):
    """Einfacher Chat-Bot mit global spezifiziertem Client, Modell und Historie"""
    global client
    global model_name
    global messages
    
    messages.append({"role": "user", "content": prompt})
    response = client.chat(model=model_name, messages=messages)
    messages.append({"role": "assistant", "content": response.message.content})
    display(Markdown(response.message.content))
```
